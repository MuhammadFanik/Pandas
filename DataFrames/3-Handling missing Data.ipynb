```python
import numpy as np
import pandas as pd
```

## None:Pythonic missing data

```python
vals1 = np.array([1, None, 3, 4])
vals1
```

```python
# The use of Python objects in an array also means that if you perform aggregations like sum() or min() across an array with a None value, you will generally get an error

# vals1.sum() --> Error
```

## NaN:Missing numerical data
NaN is specifically a floating-point value; there is no equivalent NaN value for integers, strings, or other types

```python
vals2 = np.array([1, np.nan, 3, 4])
vals2
```

```python
# Regardless of the operation, the result of arithmetic with NaN will be another NaN
print(1 + np.nan)
print(7 * np.nan)
print(vals2.sum())  # Aggregates over Nan will not result in any error, but they are not useful
print(vals2.min())

# This will ignore the missing values
print(np.nansum(vals2))
print(np.nanmin(vals2))
```

## NaN and None in Pandas
NaN and None both have their place, and Pandas is built to handle the two of them nearly interchangeably, converting between them where appropriate

```python
pd.Series([1, np.nan, 2, None])
```

# Operating on Null Values

```python
# isnull() --> Generate a Boolean mask indicating missing values
# notnull() --> Opposite of isnull()
# dropna() --> Return a filtered version of the data
# fillna () --> Return a copy of the data with missing values filled or imputed
```

```python
data = pd.Series([1, np.nan, 'hello', None])
data
```

```python
data.isnull()
```

### Dropping null values

#### Dropna() on Series

```python
data.dropna()
```

#### Dropna() for dataframe
We cannot drop single values from a DataFrame; we can only drop full rows or full columns.

```python
df = pd.DataFrame([[1, np.nan, 2],
                  [2, 3, 5],
                  [np.nan, 4, 6]])
df
```

```python
# By default dropna() will remove any row which has missing values
df.dropna()
```

```python
# Alternatively, you can drop NA values along a different axis, axis=1 drops all columns containing a null value
df.dropna(axis=1)
```

### Filling null values

#### Filling null values in Series

```python
data = pd.Series([1, np.nan, 2, None, 3], index=['a', 'b', 'c', 'd', 'e'])
data
```

```python
# We can fill NA entries with a single value, such as zero
data.fillna(0)
```

```python
# We can specify a forward-fill to propagate the previous value forward
data.ffill()    # forward-fill
```

```python
# Or we can specify a back-fill to propagate the next values backward
data.bfill()    # backward-fill
```

#### Filling null values in Dataframe
We can also specify an axis along which the fills take place

```python
df
```

```python
df.ffill(axis=1)
```
