```python
import numpy as np
import pandas as pd
```

```python
def make_df(cols, ind):
    """Quickly make a DataFrame"""
    data = {c: [str(c) + str(i) for i in ind]
    for c in cols}
    return pd.DataFrame(data, ind)
```

```python
make_df('ABC', range(3))
```

# pd.concat
Pandas concatenation preserves indices, even if the result will have duplicate indices!

### simple concatenation of Series

```python
ser1 = pd.Series(['A', 'B', 'C'], index=[1, 2, 3])
ser2 = pd.Series(['D', 'E', 'F'], index=[4, 5, 6])
```

```python
ser1
```

```python
ser2
```

```python
pd.concat([ser1, ser2])
```

### concatenation of dataframes
By default, the concatenation takes place row-wise within the DataFrame

```python
df1 = make_df('AB', [1, 2])
df2 = make_df('AB', [3, 4])
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2])
```

pd.concat allows specification of an axis along which concatenation will take place. Consider the following example

```python
df3 = make_df('AB', [0, 1])
df4 = make_df('CD', [0, 1])
```

```python
df3
```

```python
df4
```

```python
pd.concat([df3, df4], axis=1)
```

### Ignoring the dupliccate index
Sometimes the index itself does not matter, and you would prefer it to simply be ignored. You can specify this option using the ignore_index flag. With this set to True, the concatenation will create a new integer index for the resulting Series

```python
x = make_df('AB', [0, 1])
y = make_df('AB', [2, 3])
```

```python
print(x)
print(y)
```

```python
x.index = y.index
```

```python
pd.concat([x, y], ignore_index=True)
```

## Concatenation with Joins
In the simple examples we just looked at, we were mainly concatenating DataFrames with shared column names. In practice, data from different sources might have different sets of column names, and pd.concat offers several options in this case. Consider the concatenation of the following two DataFrames, which have some (but not all!) columns in common

```python
df5 = make_df('ABC', [1, 2])
df6 = make_df('BCD', [3, 4])
```

```python
df5
```

```python
df6
```

```python
pd.concat([df5, df6], join="inner")
```
