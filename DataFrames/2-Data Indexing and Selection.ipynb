# Data Selection in Series

### Series as a Dict

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0, 1.25, 1.5], index=['a', 'b', 'c', 'd', 'e', 'f'])
data
```

```python
data['d']
```

```python
# We can also use dictionary-like Python expressions and methods to examine the keys/indices and values
print('b' in data)
```

```python
# Just as you can extend a dictionary by assigning to a new key, you can extend a Series by assigning to a new index value
data['g'] = 1.75    # Adding g index in the data
data
```

```python
# Slicing by explicit index
data['a':'e']
```

```python
# Slicing by implicit Integer index
data[1:5]
```

```python
# Masking
data[(data >= 1.25)]
```

```python
# Fancy Indexing
data[['a', 'e']]
```

#### Indexers: loc/iloc
These slicing and indexing conventions can be a source of confusion. For example, if your Series has an explicit integer index, an indexing operation such as data[1] will use the explicit indices, while a slicing operation like data[1:3] will use the implicit Python-style index.

```python
data = pd.Series(['a', 'b', 'c'], index=[1, 3, 5])
data
```

```python
# Explicit index when Indexing
data[1]
```

```python
# Implicit index when Indexing
data[1:]
```

### Loc
Allows indexing and slicing that always references the explicit index

```python
data.loc[1]
```

```python
data.loc[1:3]
```

### iloc
Allows indexing and slicing that always references the implicit Python-style index

```python
data.iloc[1]
```

```python
data.iloc[1:3]
```

# Data Selection in DataFrame

### Dataframe as a dictionary

```python
pop_dict = {'California':12345,
            'New York':12346,
            'Texas':147895,
            'Florida':968574}

area_dict = {'California': 423967,
            'New York': 141297,
            'Texas': 695662,
            'Florida': 170312}

population_series = pd.Series(pop_dict)
area_series = pd.Series(area_dict)
```

```python
states = pd.DataFrame({"population": population_series,
                       "area": area_series})
states
```

```python
# Accessing a single column
states["population"]
```

```python
# Add a new column
states["density"] = (states["population"] / states["area"]) * 100
states
```

### Dataframe as a 2-d array

```python
states.values
```

```python
# Transpose
states.T
```

```python
states
```

```python
# iloc
states.iloc[:3, :2]
```

```python
# loc
states.loc[:"Texas", ["area", "density"]]
```

### Additional indexing conventions

While indexing refers to columns, slicing refers to rows

```python
states
```

```python
# Slicing
states["New York": "Texas"]
```

```python
# Such slices can also refer to rows by number rather than by index:
states[1:3]
```

```python
# Similarly, direct masking operations are also interpreted row-wise rather than column-wise
states[states.density > 20]
```
