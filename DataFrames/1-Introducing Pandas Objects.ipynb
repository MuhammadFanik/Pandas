```python
import numpy as np
import pandas as pd
```

### Dataframe as a generalized NumPy array

```python
pop_dict = {'California':12345,
            'New York':12346,
            'Texas':147895,
            'Florida':968574}

area_dict = {'California': 423967,
            'New York': 141297,
            'Texas': 695662,
            'Florida': 170312}

population_series = pd.Series(pop_dict)
area_series = pd.Series(area_dict)
```

```python
print(population_series)
print("\n")
print(area_series)
```

```python
# Constructing a Dataframe from the above information
states = pd.DataFrame({"population": population_series,
                       "area": area_series})
states
```

#### Index

```python
# Index - It gives access to the index labels
states.index
```

#### Columns

```python
# Columns - it is an index object holding the column labels
states.columns
```

### Dataframe as a specialized Dictionary

```python
states["area"]
```

# Constructing Dataframe Objects

### 1- From a Single Series objects

```python
# A DataFrame is a collection of Series objects, and a single column DataFrame can be constructed from a single Series:
pd.DataFrame(population_series, columns=["population"])
```

### 2 - From a list of Dicts

```python
# Any list of dictionaries can be made into a DataFrame. We’ll use a simple list comprehension to create some data:
data = [{'a': i, 'b':i*2} for i in range(5)]
pd.DataFrame(data)
```

### 3 - From a 2-D Numpy array

```python
c = pd.DataFrame(np.round(np.random.rand(3, 2), 2), columns=["foo", "bar"], index=[10, 20, 30])
c
```
