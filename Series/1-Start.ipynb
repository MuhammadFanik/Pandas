# Creating a Pandas Series

#### A Pandas Series is a 1-D array of indexed data. It can be created from a list or array

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0, 1.25])
data
```

```python
# Values - The values are simply a numpy array
data.values
```

```python
# Index -
data.index
```

```python
# Accessing Data
print(data[0])     # Accessing the first element
print(data[2])     # Accessing the third element
print(data[1:3])   # Accessing elements starting from 1 (inclusive) till 3 (exclusive)
```

### Series with non-integer indexes

```python
data1 = pd.Series([10, 20, 30, 40, 50], index=["Sam", "Simon", "b", "z", "Umar"])
data1
```

```python
data1["Simon":"z"]
```

### Series creation from Dictionary

```python
pop_dict = {'California':12345,
            'New York':12346,
            'Texas':147895,
            'Florida':968574}
population = pd.Series(pop_dict)
```

```python
population
```

```python
population["Texas"]
```

# Constructing Series Objects

### 1 - Data can be a numpy array / list where the index defaults to an integer sequence

```python
a = pd.Series([1, 2, 3, 4, 4])
a
```

### 2 - Data can be a scalar, which is repeated to fill the specified index

```python
b = pd.Series(2, index=[100, 200, 300, 400, 500])
b
```

### 3 - Data can be a dictionary, in which index defaults to dictionary keys

```python
c = pd.Series({2:'a', 1:'b', 3:'c'})
c
```

### In each case, the index can be explicitly set if a different result is preferred

```python
# Values with index 1 and 2 will only be shown
pd.Series({2:'a', 1:'b', 3:'c'}, index=[1, 2])
```
